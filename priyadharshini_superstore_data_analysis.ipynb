**task 1 resample daily sales ito weekly aggregate sales**

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Ensure the date column is in datetime format
df_superstore["Order Date"] = pd.to_datetime(df_superstore["Order Date"])

# 2. Set index, resample weekly ('W'), aggregate with sum, and plot
weekly_sales = (
    df_superstore.set_index("Order Date")["Sales"].resample("W").sum()
)

weekly_sales.plot(
    kind="line",
    figsize=(10, 5),
    title="Weekly Sales Trend",
    xlabel="Date",
    ylabel="Total Sales",
)

plt.show()
```

**task 2 calculate and plot 7 days moving average of sales**

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Ensure date index and aggregate sales by day
df_superstore["Order Date"] = pd.to_datetime(df_superstore["Order Date"])
daily_sales = df_superstore.set_index("Order Date")["Sales"].resample("D").sum()

# 2. Calculate 7-day rolling mean
daily_sales_7d_avg = daily_sales.rolling(window=7).mean()

# 3. Plot daily sales alongside 7-day rolling average
plt.figure(figsize=(12, 6))
plt.plot(daily_sales, label="Daily Sales", alpha=0.4, color="gray")
plt.plot(
    daily_sales_7d_avg,
    label="7-Day Rolling Average",
    color="blue",
    linewidth=2,
)

plt.title("Daily Sales with 7-Day Rolling Average")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

**task 3 calculate period-over-period daily growth using**

```python
import pandas as pd

# 1. Ensure dates are sorted and aggregated daily
df_superstore["Order Date"] = pd.to_datetime(df_superstore["Order Date"])
daily_df = (
    df_superstore.groupby("Order Date")["Sales"]
    .sum()
    .resample("D")
    .sum()
    .reset_index()
)

# 2. Create Lag-1 column using shift(1)
daily_df["Sales_Prev_Day"] = daily_df["Sales"].shift(1)

# 3. Calculate daily percentage change
# Option A: Manual calculation using the lagged column
daily_df["Sales_Pct_Change"] = (
    (daily_df["Sales"] - daily_df["Sales_Prev_Day"])
    / daily_df["Sales_Prev_Day"]
) * 100

# Option B: Built-in pandas method (equivalent result)
# daily_df['Sales_Pct_Change'] = daily_df['Sales'].pct_change() * 100

print(daily_df.head())
```

**task 2  Matplotlib 3D surface plot of synthetic scales - profit surface**

```python
import matplotlib.pyplot as plt
import numpy as np

# Define coordinate ranges
x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)

# Create a 2D coordinate grid
X, Y = np.meshgrid(x, y)

# Define a smooth height function Z = f(X, Y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R) / (R + 1e-5)  # Ripple surface with peak at origin

# Setup 3D plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot the 3D surface
surf = ax.plot_surface(
    X, Y, Z,
    cmap='viridis',       # Smooth color gradient
    edgecolor='none',     # Remove mesh line wireframes for a smooth look
    antialiased=True,     # Smooth out edge pixelation
    rstride=1,            # Row sampling stride
    cstride=1             # Column sampling stride
)

# Axis labels and formatting
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
ax.set_title('Smooth 3D Surface Plot')

# Add color bar legend
cbar = fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)
cbar.set_label('Z Value Scale')

plt.show()
```

**task 6 reactive dropdown filter(region selector)**

```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output

# 1. Load the Superstore dataset directly from a public URL
df = pd.read_csv("samplesuperstore.csv")

# 2. Set up the dropdown selector widget
region_dropdown = widgets.Dropdown(
    options=["All"] + list(df["Region"].unique()),
    value="All",
    description="Region:",
    disabled=False
)

# 3. Create an output container to handle the printed output cleanly
output_area = widgets.Output()

# 4. Define the reactive function that runs when the dropdown changes
def filter_and_print_region(change):
    selected_region = change["new"]

    with output_area:
        # Clear previous printouts when a new region is selected
        clear_output(wait=True)

        if selected_region == "All":
            filtered_df = df
        else:
            filtered_df = df[df["Region"] == selected_region]

        print(f"==================================================")
        print(f" SELECTED REGION: {selected_region}")
        print(f" Total Matching Records: {len(filtered_df):,}")
        print(f"==================================================\n")

        # Print the first 10 rows of the filtered dataset
        display(filtered_df.head(10))

# 5. Connect the dropdown change event to the function
region_dropdown.observe(filter_and_print_region, names="value")

# 6. Display the dropdown and output in Google Colab
display(region_dropdown)
display(output_area)

# Trigger initial display for default "All" selection
with output_area:
    print(f"==================================================")
    print(f" SELECTED REGION: All")
    print(f" Total Matching Records: {len(df):,}")
    print(f"==================================================\n")
    display(df.head(10))
```

**task 10 multi-component KPI card & category pie chart layout.**

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Load dataset
df = pd.read_csv("samplesuperstore.csv")

# Calculate KPI values
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()

# Aggregate Category Sales for Pie Chart
category_sales = df.groupby("Category")["Sales"].sum().reset_index()

# Create a 2x3 subplot grid (Row 1: KPI Cards, Row 2: Pie Chart across all columns)
fig = make_subplots(
    rows=2, cols=3,
    specs=[
        [{"type": "indicator"}, {"type": "indicator"}, {"type": "indicator"}],
        [{"type": "pie", "colspan": 3}, None, None]
    ],
    vertical_spacing=0.15
)

# 1. Total Sales KPI Card
fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_sales,
        number={"prefix": "$", "valueformat": ",.0f"},
        title={"text": "Total Sales"}
    ),
    row=1, col=1
)

# 2. Total Profit KPI Card
fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_profit,
        number={"prefix": "$", "valueformat": ",.0f"},
        title={"text": "Total Profit"}
    ),
    row=1, col=2
)

# 3. Total Orders KPI Card
fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_orders,
        number={"valueformat": ",d"},
        title={"text": "Total Orders"}
    ),
    row=1, col=3
)

# 4. Category Pie Chart
fig.add_trace(
    go.Pie(
        labels=category_sales["Category"],
        values=category_sales["Sales"],
        hole=0.4,  # Creates a modern donut chart appearance
        textinfo="label+percent",
        hoverinfo="label+value+percent"
    ),
    row=2, col=1
)

# Layout Styling
fig.update_layout(
    title_text="Superstore Overview: Key Performance Indicators & Sales Breakdown",
    title_x=0.5,
    template="plotly_white",
    height=700
)

fig.show()
```
